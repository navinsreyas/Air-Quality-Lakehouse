```sql
CREATE SCHEMA IF NOT EXISTS workspace.aq_dlt;
SHOW SCHEMAS IN workspace;

SELECT COUNT(*) AS bronze_rows
FROM workspace.aq_dlt.bronze_observations;

SELECT *
FROM workspace.aq_dlt.bronze_observations
LIMIT 10;


SELECT *
FROM workspace.aq_dlt.bronze_observations
LIMIT 10;



DESCRIBE TABLE workspace.aq_dlt.bronze_observations;

SELECT COUNT(*) AS silver_rows
FROM workspace.aq_dlt.silver_observations;


SELECT *
FROM workspace.aq_dlt.silver_observations
LIMIT 20;


SELECT
    obs_date,
    hour_1based,
    obs_hour,
    obs_time
FROM workspace.aq_dlt.silver_observations
WHERE hour_1based = 1
LIMIT 10;



SELECT *
FROM workspace.aq_dlt.silver_observations
WHERE site_id = 33
  AND parameter_code = 'OZONE'
  AND obs_date = '2022-01-01'
  AND hour_1based = 2;

SELECT
    SUM(CASE WHEN site_id IS NULL THEN 1 ELSE 0 END) AS missing_sites,
    SUM(CASE WHEN obs_date IS NULL THEN 1 ELSE 0 END) AS missing_dates,
    SUM(CASE WHEN hour_1based NOT BETWEEN 1 AND 24 THEN 1 ELSE 0 END) AS invalid_hours,
    SUM(CASE WHEN value IS NULL THEN 1 ELSE 0 END) AS missing_values,
    SUM(CASE WHEN value_clean < 0 THEN 1 ELSE 0 END) AS negative_clean_values
FROM workspace.aq_dlt.silver_observations;



SELECT
    value,
    value_clean,
    below_detection
FROM workspace.aq_dlt.silver_observations
WHERE below_detection = true
LIMIT 20;


SELECT *
FROM workspace.aq_dlt.silver_observations
WHERE site_id = 33
  AND parameter_code = 'OZONE'
  AND obs_date = '2022-01-01'
  AND hour_1based = 2;



SELECT
    SUM(CASE WHEN site_id IS NULL THEN 1 ELSE 0 END) AS missing_sites,
    SUM(CASE WHEN obs_date IS NULL THEN 1 ELSE 0 END) AS missing_dates,
    SUM(CASE WHEN hour_1based NOT BETWEEN 1 AND 24 THEN 1 ELSE 0 END) AS invalid_hours,
    SUM(CASE WHEN value IS NULL THEN 1 ELSE 0 END) AS missing_values,
    SUM(CASE WHEN value_clean < 0 THEN 1 ELSE 0 END) AS negative_clean_values
FROM workspace.aq_dlt.silver_observations;

SELECT
    value,
    value_clean,
    below_detection
FROM workspace.aq_dlt.silver_observations
WHERE below_detection = true
LIMIT 20;


DESCRIBE TABLE workspace.aq_silver.dim_station;


SELECT *
FROM workspace.aq_silver.dim_station
LIMIT 10;


SELECT
    site_id,
    COUNT(*) AS rows_per_site
FROM workspace.aq_silver.dim_station
GROUP BY site_id
HAVING COUNT(*) > 1;


SELECT COUNT(*) AS gold_daily_rows
FROM workspace.aq_dlt.gold_daily_site_summary;

SELECT *
FROM workspace.aq_dlt.gold_daily_site_summary
LIMIT 20;



SELECT
    site_id,
    obs_date,
    parameter_code,
    COUNT(*) AS row_count
FROM workspace.aq_dlt.gold_daily_site_summary
GROUP BY
    site_id,
    obs_date,
    parameter_code
HAVING COUNT(*) > 1;



SELECT
    MIN(completeness_pct) AS min_completeness,
    MAX(completeness_pct) AS max_completeness,
    ROUND(AVG(completeness_pct), 2) AS avg_completeness
FROM workspace.aq_dlt.gold_daily_site_summary;


SELECT
    region,
    ROUND(AVG(daily_avg), 2) AS mean_pm25
FROM workspace.aq_dlt.gold_daily_site_summary
WHERE parameter_code = 'PM2.5'
GROUP BY region
ORDER BY region;
```
