```sql
ALTER STREAMING TABLE workspace.aq_dlt.silver_observations
ALTER COLUMN obs_time COMMENT
'Observation timestamp constructed from the API date and corrected 1-based hour field, so Hour 1 maps to midnight.';

ALTER STREAMING TABLE workspace.aq_dlt.silver_observations
ALTER COLUMN value COMMENT
'Raw measurement value from the NSW source. Preserved for provenance before parameter-specific analytical cleaning.';

ALTER STREAMING TABLE workspace.aq_dlt.silver_observations
ALTER COLUMN below_detection COMMENT
'For pollutant parameters only, true when the raw reading is below zero and treated as a below-detection measurement rather than a sensor error. NULL for weather parameters.';

ALTER STREAMING TABLE workspace.aq_dlt.silver_observations
ALTER COLUMN value_clean COMMENT
'Analytical measurement value. Negative pollutant readings are floored at zero while legitimate weather values, including sub-zero temperatures, are preserved unchanged.';

ALTER STREAMING TABLE workspace.aq_dlt.silver_observations
ALTER COLUMN parameter_code COMMENT
'Measurement parameter code: PM2.5, PM10, NO2, OZONE, TEMP, HUMID, WSP, WDR or RAIN.';

ALTER STREAMING TABLE workspace.aq_dlt.silver_observations
ALTER COLUMN _source_file COMMENT
'Path of the raw JSON file from which the observation was ingested by Auto Loader.';


ALTER TABLE workspace.aq_gold.hourly_wide
ALTER COLUMN obs_time COMMENT
'Station-hour observation timestamp and part of the table grain together with site_id.';


ALTER TABLE workspace.aq_gold.hourly_wide
ALTER COLUMN pm25 COMMENT
'PM2.5 analytical measurement aligned to this station-hour.';


ALTER TABLE workspace.aq_gold.hourly_wide
ALTER COLUMN temp_c COMMENT
'Temperature in degrees Celsius. Legitimate sub-zero source measurements are preserved.';

ALTER TABLE workspace.aq_gold.hourly_wide
ALTER COLUMN humidity COMMENT
'Analytical relative humidity. Source values outside 0 to 100 remain preserved in Silver but are represented as NULL in this table.';


ALTER TABLE workspace.aq_gold.hourly_wide
ALTER COLUMN wind_dir COMMENT
'Wind direction in degrees from 0 to 360. This is a circular variable and is not interpreted using ordinary arithmetic means or linear correlation.';

ALTER TABLE workspace.aq_gold.hourly_wide
ALTER COLUMN wind_dir COMMENT
'Wind direction in degrees from 0 to 360. This is a circular variable and is not interpreted using ordinary arithmetic means or linear correlation.';


ALTER TABLE workspace.aq_gold.hourly_wide
ALTER COLUMN rainfall COMMENT
'Rainfall measurement aligned to the same station-hour. Rainfall coverage is lower than wind coverage, so rainfall analyses use paired observations only.';


SELECT
    table_schema,
    table_name,
    column_name,
    comment
FROM workspace.information_schema.columns
WHERE table_schema IN ('aq_dlt', 'aq_gold')
  AND table_name IN (
      'silver_observations',
      'hourly_wide'
  )
  AND comment IS NOT NULL
ORDER BY table_schema, table_name, ordinal_position;
```
