```sql
CREATE OR REPLACE TABLE workspace.aq_dlt.merge_demo AS
SELECT *
FROM workspace.aq_dlt.silver_observations
WHERE obs_date BETWEEN '2024-06-01' AND '2024-06-07';

SELECT COUNT(*) AS before_merge
FROM workspace.aq_dlt.merge_demo;

CREATE OR REPLACE TEMP VIEW merge_source AS
SELECT *
FROM workspace.aq_dlt.silver_observations
WHERE obs_date BETWEEN '2024-06-01' AND '2024-06-07';


MERGE INTO workspace.aq_dlt.merge_demo AS target
USING merge_source AS source
ON target.site_id = source.site_id
AND target.parameter_code = source.parameter_code
AND target.obs_time = source.obs_time
WHEN MATCHED THEN UPDATE SET *
WHEN NOT MATCHED THEN INSERT *;


SELECT COUNT(*) AS after_first_merge
FROM workspace.aq_dlt.merge_demo;


SELECT COUNT(*) AS after_second_merge
FROM workspace.aq_dlt.merge_demo;



DROP TABLE IF EXISTS workspace.aq_dlt.merge_demo;
```
