```sql
SELECT
    (SELECT COUNT(*) FROM workspace.aq_dlt.bronze_observations) AS bronze_rows,
    (SELECT COUNT(*) FROM workspace.aq_dlt.silver_observations) AS silver_rows,
    (SELECT COUNT(*) FROM workspace.aq_dlt.gold_daily_site_summary) AS daily_gold_rows,
    (SELECT COUNT(*) FROM workspace.aq_dlt.gold_regional_trend) AS regional_gold_rows;

SELECT
    parameter_code,
    COUNT(*) AS rows
FROM workspace.aq_dlt.silver_observations
GROUP BY parameter_code
ORDER BY parameter_code;


SELECT
    parameter_code,
    COUNT(*) AS rows,
    ROUND(MIN(value_clean), 3) AS min_value,
    ROUND(MAX(value_clean), 3) AS max_value,
    ROUND(AVG(value_clean), 3) AS mean_value
FROM workspace.aq_dlt.silver_observations
WHERE parameter_code IN ('TEMP', 'HUMID', 'WSP', 'RAIN')
GROUP BY parameter_code
ORDER BY parameter_code;


SELECT
    COUNT(*) AS rows,
    MIN(value_clean) AS min_wdr,
    MAX(value_clean) AS max_wdr
FROM workspace.aq_dlt.silver_observations
WHERE parameter_code = 'WDR';


SELECT
    COUNT(*) AS negative_temperature_rows,
    MIN(value) AS minimum_temperature,
    MIN(value_clean) AS minimum_clean_temperature
FROM workspace.aq_dlt.silver_observations
WHERE parameter_code = 'TEMP'
  AND value < 0;


SELECT
    parameter_code,
    COUNT(*) AS rows,
    COUNT(DISTINCT site_id) AS stations,
    MIN(obs_date) AS first_date,
    MAX(obs_date) AS last_date
FROM workspace.aq_dlt.silver_observations
WHERE parameter_code IN (
    'TEMP',
    'HUMID',
    'WSP',
    'WDR',
    'RAIN'
)
GROUP BY parameter_code
ORDER BY parameter_code;



SELECT
    COUNT(*) AS humidity_over_100,
    ROUND(
        100.0 * COUNT(*) /
        (
            SELECT COUNT(*)
            FROM workspace.aq_dlt.silver_observations
            WHERE parameter_code = 'HUMID'
        ),
        4
    ) AS pct_humidity_over_100
FROM workspace.aq_dlt.silver_observations
WHERE parameter_code = 'HUMID'
  AND value_clean > 100;


SELECT
    site_id,
    obs_time,
    value,
    value_clean
FROM workspace.aq_dlt.silver_observations
WHERE parameter_code = 'HUMID'
  AND value_clean > 100
ORDER BY value_clean DESC
LIMIT 20;


SELECT
    site_id,
    COUNT(*) AS rows_over_100,
    MIN(value_clean) AS min_over_100,
    MAX(value_clean) AS max_over_100
FROM workspace.aq_dlt.silver_observations
WHERE parameter_code = 'HUMID'
  AND value_clean > 100
GROUP BY site_id
ORDER BY rows_over_100 DESC;
```
